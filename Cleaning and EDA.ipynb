<h2 style="text-align:center; color:#1f77b4;">
CLEANING OF THE DATASET
</h2>

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("amazon.csv")
df.info()
```

```python
df[['discounted_price', 'actual_price', 'discount_percentage', 'rating', 'rating_count']].head(10)
```

```python
df['discounted_price'] = (
    df['discounted_price']
    .str.replace('₹', '', regex = False)
    .str.replace(',', '', regex = False)
    .astype(float)
)
df['discounted_price'].head(10)
```

```python
df['actual_price'] = (
    df['actual_price']
    .str.replace('₹', '', regex = False)
    .str.replace(',', '', regex = False)
    .astype(float)
)
df['actual_price'].head(10)
```

```python
df['discount_percentage'] = (
    df['discount_percentage']
    .str.replace('%', '', regex = False)
    .astype(float)
)
df['discount_percentage'].head(10)
```

```python
df['rating'].value_counts().head(10)
```

```python
df['rating'].unique()
```

```python
df['rating'] = pd.to_numeric(df['rating'], errors = 'coerce')
df['rating'].head()
```

```python
df['rating_count'].head(10)
```

```python
df['rating_count'] = (df['rating_count'].str.replace(',', '', regex = False))
df['rating_count'] = pd.to_numeric(df['rating_count'], errors = 'coerce')
df['rating_count'].head(10)
```

```python
df.info()
```

```python
df[['rating', 'rating_count']].isna().sum()
```

```python
cols_to_drop = ['img_link', 'product_link', 'review_content', 'review_title', 'review_id', 'user_id', 'user_name']
df = df.drop(columns = cols_to_drop)
df.info()
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
df['product_id'].duplicated().sum()
```

```python
df = df.drop_duplicates()
df = df.drop_duplicates(subset = 'product_id')
df.duplicated().sum()
df['product_id'].duplicated().sum()
```
